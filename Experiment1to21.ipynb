```python
import numpy as np

# 4x4 matrix: rows represent students, columns represent subjects
student_scores = np.array([
    [85, 90, 78, 88],  # Student 1
    [92, 85, 80, 75],  # Student 2
    [78, 88, 92, 84],  # Student 3
    [88, 95, 85, 90]   # Student 4
])

# Subject names in column order
subjects = ["Math", "Science", "English", "History"]

# Calculate average score for each subject (column-wise)
average_scores = np.mean(student_scores, axis=0)

# Find the subject with the highest average score
highest_index = np.argmax(average_scores)
highest_subject = subjects[highest_index]
highest_average = average_scores[highest_index]

# Display the results
for subject, avg in zip(subjects, average_scores):
    print(f"{subject}: {avg:.2f}")

print(f"\nSubject with the highest average: {highest_subject}")
print(f"Highest average score: {highest_average:.2f}")
```

```python
from google.colab import files
import numpy as np
import pandas as pd
import scipy.stats as stats

uploaded = files.upload()

file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

print('File Successfully Uploaded!')
print(df.head())
```

```python
# Separate data for Drug and Placebo groups based on 'Reduction' column
drug_group = df[df['Group'] == 'Drug']['Reduction']
placebo_group = df[df['Group'] == 'Placebo']['Reduction']


# Function to calculate 95% Confidence Interval
def calculate_95_ci(data):
  n = len(data)
  mean = np.mean(data)
  sem = stats.sem(data)  # Standard Error of Mean
  ci = stats.t.interval(0.95, df=n - 1, loc=mean, scale=sem)
  return mean, ci, n


# Calculate for Drug Group
drug_mean, drug_ci, drug_n = calculate_95_ci(drug_group)

# Calculate for Placebo Group
placebo_mean, placebo_ci, placebo_n = calculate_95_ci(placebo_group)

print('--- 95% Confidence Interval Results ---')
print(f'New Drug Group (n={drug_n}):')
print(f'  Mean Reduction: {drug_mean:.2f} mmHg')
print(f'  95% Confidence Interval: {drug_ci}\n')

print(f'Placebo Group (n={placebo_n}):')
print(f'  Mean Reduction: {placebo_mean:.2f} mmHg')
print(f'  95% Confidence Interval: {placebo_ci}')
```

```python
from google.colab import files
import pandas as pd

uploaded = files.upload()

file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

print('File Successfully Uploaded!')
print(df.head())
```

```python
# Conversion rate summary for both designs
conv_summary = df.groupby('Design')['Converted'].agg(
    Total_Visitors='count', Total_Converted='sum', Conversion_Rate='mean'
)

print('--- Conversion Rate Summary ---')
print(conv_summary)
```

```python
from statsmodels.stats.proportion import proportions_ztest

# Success counts and sample sizes for Design A and B
successes = df.groupby('Design')['Converted'].sum()
nobs = df.groupby('Design')['Converted'].count()

# Perform Two-Proportion Z-Test
z_stat, p_value = proportions_ztest(
    [successes['A'], successes['B']], [nobs['A'], nobs['B']]
)

print('--- A/B Testing Hypothesis Test Results ---')
print(f'Z-Statistic: {z_stat:.4f}')
print(f'P-Value: {p_value:.6f}')

# Conclusion based on alpha = 0.05
alpha = 0.05
if p_value < alpha:
  print(
      '\nConclusion: Reject the null hypothesis. There is a STATISTICALLY'
      ' SIGNIFICANT difference in conversion rates between Website Design A and'
      ' Design B.'
  )
else:
  print(
      '\nConclusion: Fail to reject the null hypothesis. There is NO'
      ' statistically significant difference in conversion rates between'
      ' Website Design A and Design B.'
  )
```
